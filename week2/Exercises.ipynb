```python
"""
Write a Python code block that inputs numbers into a list. Print
the largest, smallest, the sum, and the average of the numbers.
Count occurrences of a specific number in the list.
"""
my_list = []
n = int(input("enter the number of of elements you want in the list: "))
for i in range(n):
    entry = int(input(f"enter element no.{i+1}: "))
    my_list.append(entry)
largest_num = max(my_list)
sum_num = sum(my_list)
average = sum_num/n
smallest = min(my_list)
a = int(input("enter the number of whose occurrences you want to count in the list: "))
occurrences = my_list.count(a)
print(f"Largest:{largest_num}, Smallest:{smallest}, Sum:{sum_num}, Average:{average} and Occurrences:{occurrences}")
```

```python
"""
Write a Python code block to create a tuple with five elements.
Try to change one of the elements and handle the error that
occurs. Print a message that explains why the error occurred.
"""
my_tuple = ("Alpha", "Beta", "Gamma", "Delta", "Epsilon")
print(f"Original Tuple: {my_tuple}")
try:
    my_tuple[1] = "Zeta"
except TypeError as error:
    print(f"Error Type: {type(error).__name__}")
    print(f"System Message: {error}")
    print("\nExplanation: This error occurred because tuples are 'immutable'.")
```

```python
"""
Write a Python code block that inputs a sentence from the user.
Count the frequency of each word in the sentence and store the
result in a dictionary. Prints the dictionary with words as keys
and their frequencies as values.
"""
sentence = input("Enter a sentence: ")
words = sentence.split()
word_counts = {}
for word in words:
    if word in word_counts:
        word_counts[word] += 1
    else:
        word_counts[word] = 1
print("Word Frequencies:")
print(word_counts)
```

```python
"""
Write a Python code block to input numbers into two sets.
Perform union, intersection, and difference operations on the sets
and print the results.
"""
input1 = input("Enter numbers for Set 1 (separated by space): ")
set1 = set(input1.split())

input2 = input("Enter numbers for Set 2 (separated by space): ")
set2 = set(input2.split())

union_set = set1.union(set2)
intersection_set = set1.intersection(set2)
difference_set = set1.difference(set2)

print("Set 1:", set1)
print("Set 2:", set2)
print("Union:", union_set)
print("Intersection:", intersection_set)
print("Difference:", difference_set)
```
