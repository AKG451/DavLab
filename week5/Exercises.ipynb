```python
"""
Create a CSV file called “Movies.csv” with details of 10 movies-
Movie Name, Language, Genre, Rating, Review.
a. Read CSV file into a dataframe and find the movie with the
highest rating.
b. Write the details of all “Hindi movies into a file
“HindiMovies.csv”.
"""
import pandas as pd

movies = {
    'Movie Name': ['Inception', 'Dangal', 'Parasite', 'Sholay', 'The Matrix', '3 Idiots', 'Spirited Away', 'Lagaan', 'Interstellar', 'PK'],
    'Language': ['English', 'Hindi', 'Korean', 'Hindi', 'English', 'Hindi', 'Japanese', 'Hindi', 'English', 'Hindi'],
    'Genre': ['Sci-Fi', 'Biography', 'Thriller', 'Action', 'Sci-Fi', 'Comedy', 'Animation', 'Drama', 'Sci-Fi', 'Comedy'],
    'Rating': [8.8, 8.3, 8.5, 8.1, 8.7, 8.4, 8.6, 8.1, 8.7, 8.1],
    'Review': ['Excellent', 'Inspiring', 'Masterpiece', 'Classic', 'Mind-bending', 'Hilarious', 'Beautiful', 'Epic', 'Stunning', 'Thought-provoking']
}
dfMovies = pd.DataFrame(movies)
dfMovies.to_csv('Movies.csv', index=False)
dfMovies = pd.read_csv('Movies.csv')
maxRating = dfMovies['Rating'].idxmax()
maxMovie = dfMovies.loc[maxRating, 'Movie Name']
maxMovieRating = dfMovies.loc[maxRating, 'Rating']
print(f"The movie with the highest rating is {maxMovie} with a rating of {maxMovieRating}")
dfHindiMovies = dfMovies[dfMovies['Language'] == 'Hindi']
dfHindiMovies.to_csv('HindiMovies.csv', index=False)
```
