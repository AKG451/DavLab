```python
"""
1. Generate a 3x4 NumPy array with random integers between 1
and 50.
a. Calculate and print the Mean, Median, and Standard Deviation
of the array
b. Print the Sum of all elements and the sum of each row.
c. Reshape the 3x4 array into a 2x6 array and print it.
"""
import numpy as np
random_array = np.random.randint(1,51,size=(3,4))
mean_val = np.mean(random_array)
median_val = np.median(random_array)
std_val = np.std(random_array)
print(f"Mean:{mean_val}, Median: {median_val}, Standard Deviation: {std_val}")
sum_value = np.sum(random_array)
sum_row = np.sum(random_array, axis=1)
print(f"Sum of Value:{sum_value}, Sum of Row: {sum_row}")
new_array = random_array.reshape(6,2)
print(f"{new_array}")
```
